```python
# Funksjon som bruker pathlib.path 
from pathlib import Path

def robust_reader_function(fil_vei: str | Path):
    sti = Path(fil_vei)
    try:
        #Prøver med utf-8
        file_text_content = sti.read_text(encoding="utf-8")
        encoding_used = "Utf-8"
        return (file_text_content, encoding_used)
    except UnicodeDecodeError: #Hvis utf-8 feiler prøver den med iso
        try:
            file_text_content = sti.read_text(encoding="iso-8859-1")
            encoding_used = "iso-8859-1"
            return (file_text_content, encoding_used)


#Skal returnere en tuple med (file_text_content, encoding_used) .

# Lage en "log parser function "
#Inneholder gyldige salg + uregelmessigheter

#Skal returnere en liste med Python-dictionaries hvor hver dictionary er en gyldig transaksjon...

```
